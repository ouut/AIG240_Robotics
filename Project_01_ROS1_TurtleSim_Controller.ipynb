## Project goal
Build a ROS package that controls a TurtleSim turtle using keyboard input. Your controller should accept the turtle name as an argument and publish velocity commands to the correct topic.

## Required behavior
- Package name: `proj1_turtlesim`.
- The node accepts a target turtle name, for example `turtle1` or `turtle2`.
- The controller publishes `geometry_msgs/Twist` messages.
- `w/a/s/d` provide basic motion; diagonal/combined options may be implemented with `q/e/z/c` or true multi-key input.
- The turtle stops when no valid key is pressed.


## Suggested setup
```bash
mkdir -p ~/proj1_ws/src
cd ~/proj1_ws/src
catkin_create_pkg proj1_turtlesim rospy geometry_msgs
mkdir -p proj1_turtlesim/scripts
cd ~/proj1_ws
catkin_make
source devel/setup.bash
```


## Implementation guidance
Structure your node into functions:
1. Parse the target turtle name from `sys.argv`.
2. Build the topic name: `/{turtle_name}/cmd_vel`.
3. Read keys without blocking the ROS shutdown process.
4. Convert keys to linear/angular velocity.
5. Publish zero velocity when stopping.
6. Log which turtle is being controlled.


## Testing protocol
1. Start `roscore`.
2. Launch TurtleSim.
3. Spawn at least one extra turtle using `rqt` or `rosservice`.
4. Run one controller per turtle.
5. Verify topic publication with:
```bash
rostopic echo /turtle1/cmd_vel
rostopic hz /turtle1/cmd_vel
```
