## Project brief

Your robot must follow a **1-meter square-style motion pattern** using a combination of forward motion, lateral motion, turning, and a final combined move-and-turn segment.

### Required motion pattern

Starting from an approximate initial pose:

1. **Move forward** from `(0, 0, 0°)` to approximately `(1, 0, 0°)` while facing forward.
2. **Move left sideways** from `(1, 0, 0°)` to approximately `(1, 1, 0°)` without rotating.
3. **Turn clockwise in place** from `(1, 1, 0°)` to approximately `(1, 1, -90°)`.
4. **Move right sideways** approximately 1 meter while maintaining the new orientation.
5. **Move forward while turning** for the final combined movement segment.
6. Repeat the complete pattern **two times**.

The exact final physical position may not perfectly match the target because the robot is controlled open-loop. Your job is to tune the commands, document the error, and explain what happened.

# Part C — Motion design strategy

Because this project runs in the **real world**, you need a practical control strategy.

## C1) Open-loop control

The simplest approach is **time-based open-loop control**:

- choose a linear or lateral velocity,
- estimate how long to command it to travel about 1 meter,
- choose an angular velocity,
- estimate how long to command it to rotate 90 degrees.

This approach is simple but imperfect:

- wheel slip changes real displacement,
- battery level changes response,
- floor texture affects motion,
- lateral mecanum/omni motion may drift,
- turning may not be perfectly centered,
- the robot may not execute exactly the same command every time.

That is why you must calibrate and document your observations.


## C2) Suggested calibration table

Create and fill in your own table during testing.

| Motion | Velocity command | Target | Initial time estimate | Measured result | Final tuned time |
|---|---|---:|---:|---|---:|
| Forward | `linear.x = ?` | 1.0 m | ? s | too short / too long | ? s |
| Left sideways | `linear.y = ?` | 1.0 m | ? s | drifted / close | ? s |
| Clockwise turn | `angular.z = ?` | 90° | ? s | under / over turn | ? s |
| Right sideways | `linear.y = ?` | 1.0 m | ? s | drifted / close | ? s |
| Forward + turn | `linear.x = ?`, `angular.z = ?` | combined path | ? s | describe | ? s |

Use this table as part of your final submission.


## C3) Safe tuning order

Do not run the full pattern immediately. Tune in this order:

1. Stop command.
2. Very short forward command.
3. Forward 1-meter segment.
4. Very short lateral command.
5. Lateral 1-meter segment.
6. Very short rotation command.
7. 90-degree turn.
8. Combined forward-and-turn segment.
9. One full pattern slowly.
10. Two full repetitions.

Only increase speed or duration after the robot behaves safely at lower values.

Paste the following starter implementation into `real_square_motion.py`.


````python
#!/usr/bin/env python
import rospy
from geometry_msgs.msg import Twist


class JetAutoPatternController:
    """Timed open-loop JetAuto motion controller.

    This controller publishes Twist commands to the robot base. It is intentionally
    simple so that students can focus on calibration, safety, and real-world error.
    """

    def __init__(self):
        rospy.init_node("jetauto_pattern_controller", anonymous=False)

        # Change this topic if your robot uses a different velocity command topic.
        self.cmd_topic = rospy.get_param("~cmd_topic", "/cmd_vel")
        self.pub = rospy.Publisher(self.cmd_topic, Twist, queue_size=10)

        # Safe engineering defaults. These must be tuned on the real robot.
        self.forward_speed = rospy.get_param("~forward_speed", 0.20)
        self.lateral_speed = rospy.get_param("~lateral_speed", 0.20)
        self.turn_speed = rospy.get_param("~turn_speed", -0.70)  # negative often means clockwise
        self.combo_forward_speed = rospy.get_param("~combo_forward_speed", 0.18)
        self.combo_turn_speed = rospy.get_param("~combo_turn_speed", 0.70)

        # Timing constants to be calibrated experimentally.
        self.forward_time = rospy.get_param("~forward_time", 5.0)
        self.left_time = rospy.get_param("~left_time", 5.0)
        self.turn_time = rospy.get_param("~turn_time", 2.2)
        self.right_time = rospy.get_param("~right_time", 5.0)
        self.combo_time = rospy.get_param("~combo_time", 5.5)

        self.repetitions = rospy.get_param("~repetitions", 2)
        self.rate = rospy.Rate(20)

    def publish_for_duration(self, linear_x=0.0, linear_y=0.0, angular_z=0.0, duration=1.0):
        """Publish a constant velocity command for a fixed duration, then stop."""
        msg = Twist()
        msg.linear.x = linear_x
        msg.linear.y = linear_y
        msg.angular.z = angular_z

        rospy.loginfo(
            "cmd => topic=%s, linear.x=%.3f, linear.y=%.3f, angular.z=%.3f, duration=%.2f",
            self.cmd_topic,
            linear_x,
            linear_y,
            angular_z,
            duration,
        )

        t_end = rospy.Time.now() + rospy.Duration(duration)
        while rospy.Time.now() < t_end and not rospy.is_shutdown():
            self.pub.publish(msg)
            self.rate.sleep()

        self.stop_robot()

    def stop_robot(self):
        """Send zero velocity to stop the robot."""
        self.pub.publish(Twist())
        rospy.sleep(0.3)

    def run_once(self):
        """Run one complete square-style pattern."""
        rospy.loginfo("Step 1: move forward")
        self.publish_for_duration(linear_x=self.forward_speed, duration=self.forward_time)

        rospy.loginfo("Step 2: move left sideways")
        self.publish_for_duration(linear_y=self.lateral_speed, duration=self.left_time)

        rospy.loginfo("Step 3: turn clockwise in place")
        self.publish_for_duration(angular_z=self.turn_speed, duration=self.turn_time)

        rospy.loginfo("Step 4: move right sideways")
        self.publish_for_duration(linear_y=-self.lateral_speed, duration=self.right_time)

        rospy.loginfo("Step 5: move forward while turning")
        self.publish_for_duration(
            linear_x=self.combo_forward_speed,
            angular_z=self.combo_turn_speed,
            duration=self.combo_time,
        )

    def wait_for_start(self):
        raw_input("Place robot at start pose, confirm the area is clear, then press Enter to begin...")

    def run(self):
        self.wait_for_start()

        for i in range(self.repetitions):
            rospy.loginfo("Starting pattern repetition %d / %d", i + 1, self.repetitions)
            self.run_once()

        rospy.loginfo("Pattern complete. Sending final stop command.")
        self.stop_robot()


if __name__ == "__main__":
    try:
        controller = JetAutoPatternController()
        controller.run()
    except rospy.ROSInterruptException:
        pass
````

Paste this simple keyboard test script.


````python
#!/usr/bin/env python
import sys
import termios
import tty

import rospy
from geometry_msgs.msg import Twist


def get_key():
    settings = termios.tcgetattr(sys.stdin)
    try:
        tty.setraw(sys.stdin.fileno())
        key = sys.stdin.read(1)
    finally:
        termios.tcsetattr(sys.stdin, termios.TCSADRAIN, settings)
    return key


def main():
    rospy.init_node("keyboard_twist_test")
    cmd_topic = rospy.get_param("~cmd_topic", "/cmd_vel")
    speed = rospy.get_param("~speed", 0.10)
    turn = rospy.get_param("~turn", 0.40)
    pub = rospy.Publisher(cmd_topic, Twist, queue_size=10)

    print("Keyboard test controls:")
    print("  w: forward")
    print("  s: backward")
    print("  a: left strafe")
    print("  d: right strafe")
    print("  q: turn left")
    print("  e: turn right")
    print("  space: stop")
    print("  x: exit")

    while not rospy.is_shutdown():
        key = get_key()
        msg = Twist()

        if key == "w":
            msg.linear.x = speed
        elif key == "s":
            msg.linear.x = -speed
        elif key == "a":
            msg.linear.y = speed
        elif key == "d":
            msg.linear.y = -speed
        elif key == "q":
            msg.angular.z = turn
        elif key == "e":
            msg.angular.z = -turn
        elif key == " ":
            msg = Twist()
        elif key == "x":
            pub.publish(Twist())
            break
        else:
            continue

        pub.publish(msg)
        rospy.sleep(0.2)
        pub.publish(Twist())


if __name__ == "__main__":
    main()
````
