## Project goal
Create a map of an assigned real-world area using the JetAuto robot and SLAM. Save the generated map files and analyze the map quality.

## Objectives
- Explain why mapping is important for autonomous robots.
- Use SLAM to build an occupancy grid.
- Drive the robot to cover the assigned mapping zone.
- Save `.pgm` and `.yaml` map files.
- Analyze map quality and identify errors.


## Procedure
1. Set up and inspect the robot with your group.
2. Launch SLAM using the Lab 6 workflow.
3. Slowly teleoperate through the assigned zone.
4. Watch RViz for live map coverage.
5. Avoid fast rotations and repeated passes that create duplicate walls.
6. Save the map once the zone is covered.


## Map analysis guide
In your screenshot, annotate:
- clear walls and boundaries
- open spaces
- doors or narrow passages
- furniture or obstacles
- discontinuities, duplicated walls, or missing areas

Explain whether each issue likely came from sensing, localization drift, fast movement, reflective surfaces, or dynamic obstacles.


## Assessment questions
1. What is the purpose of mapping in autonomous robots?
2. What map representation is commonly used in ROS mapping?
3. How does RViz help during mapping?
4. What can affect map accuracy? How do chairs, tables, doors, and windows appear?
5. What role do robot sensors play in building the map?
6. What files are created when a map is saved, and what does each file do?

## Submission checklist
- Saved map files from the classroom and assigned zone.
- Annotated screenshots with error analysis.
- Short video of RViz mapping with the real robot.
- Group answer file.


## Grading rubric — 20 points

| Category | Weight | Excellent | Satisfactory | Needs improvement |
|---|---:|---|---|---|
| Robot setup, safety, and teamwork | 2 pts | Robot is inspected before use, group roles are clear, safe operating speed is maintained, and the team responds appropriately to obstacles or people near the robot. | Robot is used safely with minor instructor reminders. Group participation is mostly balanced. | Safety steps are missing, the robot is driven carelessly, or teamwork is unclear. |
| SLAM launch and ROS workflow | 3 pts | Correctly launches the required robot drivers, SLAM nodes, RViz visualization, and verifies key topics/frames before mapping. | Launches the required tools but verification is incomplete or partially documented. | Cannot reliably launch SLAM, misses important nodes/topics, or depends heavily on instructor intervention. |
| Mapping coverage and driving strategy | 4 pts | Covers the assigned zone systematically using slow, smooth motion; revisits important areas without creating major duplicate walls; avoids unnecessary fast rotations. | Covers most of the assigned zone, but the path is not fully systematic or includes some movement that reduces map quality. | Coverage is incomplete, motion is too fast/erratic, or the map contains major avoidable errors. |
| Map quality | 4 pts | Final map has clear walls/boundaries, recognizable open spaces, reasonable obstacle placement, and limited drift or duplication. | Final map is usable but has some missing regions, duplicated structures, or moderate noise. | Final map is difficult to interpret, has severe drift, or does not represent the assigned area well. |
| Map saving and file organization | 2 pts | Correctly saves both `.pgm` and `.yaml` files, uses meaningful filenames, and organizes files/screenshots clearly for submission. | Saves required files but naming or organization could be clearer. | Missing map files, incorrect file formats, or files cannot be matched to the submitted work. |
| Evidence and documentation | 2 pts | Provides clear screenshots, short RViz video, command/topic evidence, and annotations that support the mapping result. | Provides most required evidence, but some screenshots, annotations, or command evidence are incomplete. | Evidence is missing, unclear, or does not demonstrate that the group performed the mapping task. |
| Map analysis and reflection | 3 pts | Clearly explains map strengths, errors, likely causes, and practical improvements using correct robotics vocabulary. | Explains the map result with some correct observations, but analysis is shallow or missing causes/improvements. | Reflection is vague, mostly descriptive, or does not connect map errors to SLAM, sensing, motion, or environment factors. |

## Minimum submission requirements

A complete submission must include:

1. Saved map files: `.pgm` and `.yaml`.
2. At least one RViz screenshot showing the generated map.
3. One annotated map screenshot identifying good regions and problem areas.
4. A short video or screen recording showing mapping in progress.
5. Written answers to the assessment questions.
6. A short reflection explaining what the group would change to improve map quality.
