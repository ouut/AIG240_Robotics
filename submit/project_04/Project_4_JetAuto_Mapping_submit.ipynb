## Map Statistics

Analysing the occupancy grid to quantify map quality.

```python
import numpy as np
from PIL import Image

img = Image.open('project4/explore.pgm')
data = np.array(img)
print(f'Dimensions : {data.shape[1]} × {data.shape[0]} pixels')
print(f'Resolution : 0.025 m/px → {data.shape[1]*0.025:.1f} m × {data.shape[0]*0.025:.1f} m')

unique, counts = np.unique(data, return_counts=True)
for v, c in zip(unique, counts):
    pct = 100*c / data.size
    label = 'OCCUPIED (wall)' if v==0 else ('FREE' if v==254 else 'FREE (explored)')
    print(f'  Value {v:>3}: {c:>8} px  ({pct:5.1f} %)  — {label}')

wall = int(np.sum(data == 0))
free = int(np.sum(data > 0))
print(f'\nWall pixels  : {wall}')
print(f'Free pixels  : {free}')
print(f'Explored area: {100*free/(wall+free):.1f} %')
```

## Assessment Questions

### 1. What is the purpose of mapping in autonomous robots?

Mapping gives an autonomous robot a spatial model of its environment so it can localise itself (know *where* it is), plan collision‑free paths, and navigate to goal locations. Without a map, the robot cannot distinguish free space from obstacles and has no reference for planning or re‑localisation. A map is also the output of SLAM — it encodes the robot's cumulative sensor observations into a persistent, reusable representation of the world.

### 2. What map representation is commonly used in ROS mapping?

ROS mapping typically uses an **occupancy grid** — a 2‑D grid where each cell stores a probability that the corresponding physical location is occupied. Common values are:
- **0 (black)**: occupied — wall, obstacle, or furniture
- **254 (white)**: free — open, traversable space
- **Intermediate grey (~205)**: unknown or free (depending on SLAM configuration)

The occupancy grid is saved as a portable greymap (`.pgm`) image paired with a `.yaml` metadata file that records resolution, origin, and threshold parameters needed to convert pixel values into occupancy probabilities.

### 3. How does RViz help during mapping?

RViz is the 3‑D visualisation tool in ROS. During mapping it:
- Displays the **live occupancy grid** as it is built, so the operator can see which areas have been mapped and which are still unknown.
- Shows the robot's **estimated pose** on the map, helping the operator detect localisation drift early.
- Overlays the **laser scan** points so the operator can verify that sensor readings align with the existing map.
- Provides a real‑time feedback loop: the operator can slow down, revisit an area, or adjust the path based on what RViz shows.

Without RViz, the operator is effectively driving blind — they have no way to assess map coverage or quality until after the map is saved.

### 4. What can affect map accuracy? How do chairs, tables, doors, and windows appear?

**Factors that reduce map accuracy:**
- **Fast rotation** — the LiDAR scan matching degrades, causing walls to smear or duplicate.
- **Slippery floors or wheel slip** — odometry errors cause localisation drift; walls appear shifted or doubled.
- **Reflective surfaces** (glass, polished metal) — the laser either passes through or bounces unpredictably, creating phantom obstacles or missing walls.
- **Dynamic obstacles** (people walking) — appear as temporary walls that vanish later, leaving ghost artefacts if not cleared.
- **Feature‑poor areas** (long featureless corridors) — scan matching has insufficient distinct features, leading to drift along the corridor axis.

**How common objects appear:**
- **Chairs and tables**: Thin leg structures appear as isolated small occupied cells (dots or short segments). Table tops at LiDAR height produce a horizontal line. If the LiDAR scans above or below table tops, the space underneath appears free.
- **Doors**: Closed doors appear as a solid wall segment. Open doors appear as a narrow gap in the wall. Glass doors may be invisible to the LiDAR.
- **Windows**: Glass is mostly transparent to the LiDAR, so windows typically appear as gaps in the wall (missing occupancy) where the laser beam passes through.

### 5. What role do robot sensors play in building the map?

The JetAuto uses a **2‑D LiDAR (RPLIDAR A1)** as its primary mapping sensor:
- The LiDAR emits laser pulses and measures time‑of‑flight to return a 360° set of range readings (typically at 5–10 Hz).
- These range readings are converted into a **laser scan** (`sensor_msgs/LaserScan`), which the SLAM node uses to update the occupancy grid — marking cells along each beam as free and the end‑point where the beam hits as occupied.

**Wheel odometry** (from the motor encoders) provides an initial pose estimate. The SLAM algorithm (gmapping) fuses odometry with laser scans to correct drift — odometry alone would accumulate unbounded error, but when corrected by scan matching against the map, it stays accurate.

Without either sensor: no laser → no obstacle detection; no odometry → no motion estimate → SLAM cannot initialise.

### 6. What files are created when a map is saved, and what does each file do?

ROS `map_server` saves two files:

1. **`explore.pgm`** (Portable Greymap):
   - A grayscale image where each pixel represents an occupancy grid cell.
   - Pixel value 0 = occupied, 254 = free, intermediate values = unknown.
   - Our map: 800 × 832 pixels at 2.5 cm/pixel → covers 20.0 m × 20.8 m.

2. **`explore.yaml`** (YAML metadata):
   - `image`: path to the PGM file.
   - `resolution`: metres per pixel (here 0.025 m/px).
   - `origin`: the real‑world coordinates `[x, y, yaw]` of the bottom‑left pixel (here `[-5, -5, 0]`, placing the map centre near the origin).
   - `negate`: whether to invert black/white semantics (0 = no inversion).
   - `occupied_thresh` / `free_thresh`: probability thresholds used to convert pixel values into the ternary occupied/free/unknown classification.

Together these two files allow any ROS node to reconstruct the metric map for navigation (e.g., in `move_base`).

## Reflection — What we would change to improve map quality

1. **Slower, more deliberate driving**: During the session, a few turns were faster than ideal. Next time we would enforce a maximum angular velocity of 0.25 rad/s during rotation, and pause for 2–3 seconds after each turn to let scan matching re‑converge before moving forward.

2. **Loop closure paths**: We drove mostly in a serpentine pattern. Adding deliberate loop closures (returning to a previously‑visited area from a different angle) would help gmapping correct accumulated drift and produce sharper wall alignment.

3. **Multi‑pass edge mapping**: After covering the interior, we would trace along all walls at a slow, constant speed and at a consistent distance (~0.5 m) to fill in thin or missing wall segments.

4. **Avoid dynamic obstacles**: A few people walked through the mapping zone, creating temporary occupied cells that disappeared later. We would wait for the area to clear before starting, and pause mapping if someone enters.

5. **Pre‑inspect the LiDAR height**: Furniture that sits below the LiDAR scan plane is invisible. We would note the LiDAR height (typically ~20 cm on the JetAuto) and identify which obstacles will or will not be detected before driving.

6. **Verify topic rates**: Before mapping next time, we would run `rostopic hz /scan` and `rostopic hz /tf` to confirm the LiDAR and transform rates are stable (≥5 Hz), as low rates cause scan‑matching gaps that appear as missing walls.

## 4. Assessment Questions

### Q1. What is the purpose of mapping in autonomous robots?

Mapping provides the robot with a **spatial model of its environment** — a persistent representation of where obstacles, walls, and free space are located. This model is essential for three core capabilities:
- **Localisation**: the robot compares current sensor readings against the map to estimate its own pose (AMCL).
- **Path planning**: a global planner (e.g. Dijkstra, A\*) uses the map to compute a collision‑free route from start to goal.
- **Obstacle avoidance**: the local planner uses the map to react to dynamic obstacles while following the global plan.

Without a map, the robot is essentially blind — it cannot plan, cannot re‑localise if it gets lost, and cannot operate autonomously in a known space.

### Q2. What map representation is commonly used in ROS mapping?

ROS uses the **occupancy grid** (`nav_msgs/OccupancyGrid`), saved as a pair of files:

| Value | Meaning | PGM pixel |
|---|---|---|
| 0 (unoccupied) → 100 (occupied) | Probability × 100 | — |
| < `free_thresh` (e.g. 0.196) | Free space | White (254) |
| > `occupied_thresh` (e.g. 0.65) | Occupied | Black (0) |
| Between thresholds | Unknown | Grey (205) |

Each cell is typically 2.5–5 cm in the real world. The `.yaml` metadata file records resolution, origin, and thresholds so `map_server` can reconstruct the metric map for navigation.

### Q3. How does RViz help during mapping?

RViz is the real‑time visual feedback loop during SLAM mapping:
- **Live occupancy grid** — the operator sees which areas have been explored (grey) and which are still unknown.
- **Laser scan overlay** — verifies that the LiDAR readings align with the existing map; misalignment signals localisation drift.
- **Robot pose** — the `base_link` → `map` transform shows whether the robot believes it is where it actually is.
- **Coverage monitoring** — the operator can spot unexplored corners and steer the robot to fill them before saving.

Without RViz, mapping is guesswork — you cannot assess map quality until after the run.

### Q4. What can affect map accuracy? How do chairs, tables, doors, and windows appear?

**Factors degrading accuracy:**
- **Fast rotations** → scan matching fails, producing smeared or duplicated walls.
- **Wheel slip / odometry error** → drift causes walls to appear shifted or doubled.
- **Reflective surfaces (glass, metal)** → LiDAR beams pass through or bounce, creating phantom obstacles or missing walls.
- **Dynamic obstacles (people)** → temporary occupied cells that later become free, leaving ghost artefacts.
- **Feature‑poor corridors** → insufficient distinct scan features → drift along the corridor axis.

**Object appearance:**
- **Chairs/desks**: thin legs appear as small isolated occupied cells; desk tops at LiDAR height create a horizontal line. Obstacles below the LiDAR plane are invisible.
- **Doors**: closed → solid wall segment; open → narrow gap. Glass doors may be invisible.
- **Windows**: glass is transparent to 905 nm LiDAR → appear as gaps in the wall.

### Q5. What role do robot sensors play in building the map?

**RPLIDAR A1 (2‑D LiDAR)**: the primary mapping sensor. Emits 360° laser pulses at ~5–10 Hz, measuring time‑of‑flight to obstacles up to ~12 m. Each scan is published as `sensor_msgs/LaserScan`. Gmapping uses these ranges to update the occupancy grid — marking cells along each ray as free and the endpoint as occupied.

**Wheel odometry** (motor encoders): provides an initial pose estimate via dead reckoning. Odometry alone drifts without bound; gmapping fuses it with laser scans through particle‑filter‑based scan matching to produce a globally consistent map and pose.

**Complementary role**: the LiDAR corrects odometry drift; odometry provides continuity between scans. Without either sensor, SLAM cannot initialise or maintain a consistent map.

### Q6. What files are created when a map is saved, and what does each file do?

`rosrun map_server map_saver` produces two files:

1. **`explore.pgm`** (Portable Greymap image)
   - Each pixel = one occupancy grid cell.
   - 0 (black) = occupied, 254 (white) = free, intermediate = unknown.
   - Our map: 800 × 832 px at 2.5 cm/px → physical area 20.0 m × 20.8 m.
   - The image is human‑readable (can be opened in any image viewer).

2. **`explore.yaml`** (YAML metadata)
   - `image`: path to the PGM file.
   - `resolution`: 0.025 metres per pixel.
   - `origin`: `[x, y, yaw]` of the bottom‑left pixel in the world frame (here `[-5, -5, 0]`, centring the map near the origin).
   - `negate`: 0 (white = free, black = occupied — no inversion).
   - `occupied_thresh` / `free_thresh`: probability cutoffs for the ternary classification.

Together they allow `map_server` to serve the map for navigation (e.g. `move_base`).
